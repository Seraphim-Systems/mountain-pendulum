# Training and Evaluation Workflow
Run baseline experiments and inspect outputs.

```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd().resolve().parent
print('Project root:', ROOT)
```

```python
# Train a baseline.
cmd = [
    sys.executable, '-m', 'src.training.train',
    '--config', 'configs/q_learning_discrete.yaml',
    '--project-root', '.'
]
print('Running:', ' '.join(cmd))
# subprocess.run(cmd, check=True)
```

```python
# Evaluate a trained checkpoint (update file name to an existing model).
eval_cmd = [
    sys.executable, '-m', 'src.training.evaluate',
    '--config', 'configs/q_learning_discrete.yaml',
    '--checkpoint', 'outputs/models/q_learning_discrete_seed42_final.npz',
    '--seed', '42',
    '--episodes', '20',
    '--project-root', '.'
]
print('Running:', ' '.join(eval_cmd))
# subprocess.run(eval_cmd, check=True)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

log_files = sorted(Path('outputs/logs').glob('*_episodes.csv'))
if not log_files:
    print('No episode CSV files found yet. Run training first.')
else:
    df = pd.read_csv(log_files[-1])
    plt.figure(figsize=(9, 4))
    plt.plot(df['episode'], df['reward'])
    plt.title('Latest run reward curve')
    plt.xlabel('Episode')
    plt.ylabel('Reward')
    plt.tight_layout()
    plt.show()
```
